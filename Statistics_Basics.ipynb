```python
#4. Implement sampling techniques like random sampling and stratified sampling

import numpy as np
from scipy import stats as st

def random_sampling(data, sample_size):
    return np.random.choice(data, size=sample_size, replace=False)

def stratified_sampling(data, strata, sample_size_per_stratum):


    if len(data) != len(strata):
        print("Error: Data and strata lists must have the same length.")
        return []

    sampled_data = []
    unique_strata = np.unique(strata)

    for stratum in unique_strata:
      # Find indices belonging to the current stratum
        stratum_indices = np.where(strata == stratum)[0]
        # Sample from the current stratum
        if len(stratum_indices) > 0: # Check if stratum_indices is not empty
            stratum_sample = np.random.choice(stratum_indices, size=min(sample_size_per_stratum, len(stratum_indices)), replace=False)
            sampled_data.extend([data[i] for i in stratum_sample])
    return sampled_data

# Example Usage:
population = np.arange(100)  # Example Population
sample_size = 10

# Random Sampling
random_sample = random_sampling(population, sample_size)
print("Random Sample:", random_sample)


# Stratified Sampling (Example with even numbered and odd numbered strata)

strata = [i % 2 for i in population] # creates a stratum for even (0) and odd (1) numbers in the population
stratified_sample = stratified_sampling(population, strata, sample_size // 2) # dividing the sample size by 2 for even distribution into the strata
print("Stratified Sample:", stratified_sample)
```

```python
#12.  Implement and compare simple random sampling and systematic sampling

import numpy as np

def simple_random_sampling(population, sample_size):

    return np.random.choice(population, size=sample_size, replace=False)

def systematic_sampling(population, sample_size):

    interval = len(population) // sample_size
    start = np.random.randint(0, interval)
    indices = np.arange(start, len(population), interval)
    return population[indices]

# Example usage
population_size = 100
population = np.arange(population_size)
sample_size = 10


simple_random_sample = simple_random_sampling(population, sample_size)
systematic_sample = systematic_sampling(population, sample_size)

print("Simple Random Sample:", simple_random_sample)
print("Systematic Sample:", systematic_sample)

print("\nComparison:")
print(f"Simple Random Sample Mean: {np.mean(simple_random_sample)}")
print(f"Systematic Sample Mean: {np.mean(systematic_sample)}")
print(f"Simple Random Sample Variance: {np.var(simple_random_sample)}")
print(f"Systematic Sample Variance: {np.var(systematic_sample)}")
```

```python
 #13.Calculate the mean, median, and mode of grouped data


import numpy as np

def calculate_grouped_mean(data, frequencies):
    weighted_sum = sum(data[i] * frequencies[i] for i in range(len(data)))
    total_frequency = sum(frequencies)
    return weighted_sum / total_frequency

def calculate_grouped_median(data, frequencies):
    total_frequency = sum(frequencies)
    cumulative_frequency = 0
    for i in range(len(data)):
        cumulative_frequency += frequencies[i]
        if cumulative_frequency >= (total_frequency + 1) / 2:
            if i == 0:
                return data[i]
            else:
                return data[i]

def calculate_grouped_mode(data, frequencies):
    mode_index = np.argmax(frequencies)
    return data[mode_index]



data = [15, 25, 35, 45, 55]
frequencies = [2, 5, 8, 4, 1]

mean = calculate_grouped_mean(data, frequencies)
median = calculate_grouped_median(data, frequencies)
mode = calculate_grouped_mode(data, frequencies)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

```python
#18. Implement Z-score normalization and explain its significance

import numpy as np

def zscore_normalize(data):

    data = np.array(data)
    mean = np.mean(data)
    std_dev = np.std(data)
    normalized_data = (data - mean) / std_dev
    return normalized_data


data = [10, 20, 30, 40, 50, 60]
normalized_data = zscore_normalize(data)
print(f"Original data: {data}")
print(f"Normalized data: {normalized_data}")


# Significance of Z-score normalization:

# 1. Centering and Scaling: Z-score normalization transforms data by subtracting the mean and dividing by the standard deviation. This centers the data around zero and scales it to have a unit standard deviation.
# 2. Feature Scaling:  In machine learning, many algorithms are sensitive to the scale of features. Z-score normalization ensures that features with larger magnitudes don't disproportionately influence the model. This can improve model performance and stability.
# 3. Outlier Detection: Z-scores can highlight outliers. Data points with a Z-score significantly greater than 3 or less than -3 are often considered outliers.
# 4. Data Comparison: When comparing datasets with different units or scales, Z-score normalization allows for direct comparison of the data points because they are transformed to a common scale.
# 5. Statistical Analysis: Z-scores are used in hypothesis testing, where we can determine the probability of observing a data point given a certain distribution.
```

```python
#24. Implement the Pearson and Spearman correlation coefficients for a dataset.

import pandas as pd

def calculate_correlation(data1, data2):

    df = pd.DataFrame({'data1': data1, 'data2': data2})
    pearson_corr = df.corr(method='pearson').iloc[0, 1]
    spearman_corr = df.corr(method='spearman').iloc[0, 1]

    return {'pearson': pearson_corr, 'spearman': spearman_corr}


data1 = [1, 2, 3, 4, 5]
data2 = [2, 4, 1, 5, 3]

correlations = calculate_correlation(data1, data2)
print(f"Pearson correlation: {correlations['pearson']}")
print(f"Spearman correlation: {correlations['spearman']}")
```
